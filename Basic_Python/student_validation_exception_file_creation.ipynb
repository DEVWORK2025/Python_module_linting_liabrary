# Student validation

```python
import sys
```

```python
def val_name(name):
    try:
        message = None
        if name.isalpha():
            flag = True
        else:
            flag= False
            raise Exception("Enter only alphabets")
    except:
        message =sys.exc_info()[1]
    return flag,message
```

```python
val_name('24d')
```

```python
def val_mobile(mob):
    try:
        message = None
        if len(str(mob))==10 and mob.isdigit()==True:
            flag= True
        else:
            flag =False
            raise Exception("Enter only digit")
    except:
         message = print(sys.exc_info()[1])
    return flag, message
```

```python
def val_marks(m):
    try:
        message = None
        if m.isdigit():
            m=float(m)
            if m >=0 and  m <=100:
                flag =  True
            else:
                flag = False
                raise Exception("number should be between 0 to 100")
        else:
            flag =  False
            raise Exception("enter only digits")
    except:
        message = sys.exc_info()[1]
    return flag , message
```

```python
f1 = open("log.txt", 'x')
```

```python
dict12 = {}
```

```python
while True:
    f1.write("Enter the name:")
    name  = input('enter the name')
    f1.write("\t"+str(name))
    name_info = val_name(name)
    f1.write(str(name_info[1]))
    ###print(name_info)
    
    if name_info[0] == True:
        break
    else:
        print(name_info[1])
        
while True:    
    mobile = input("Enter the mobile")
    mobile_info = val_mobile(mobile)
    
    if mobile_info[0] == True:
        break
    else:
        print(mobile_info[1])
        
while True:   
    marks = input("Enter the marks")
    marks_info = val_marks(marks)
    
    if marks_info[0] == True:
        break
    else:
        print(marks_info[1])

dict12.update({len(dict12)+1:{"name": name, "mobile":mobile, "marks":marks}})
```
