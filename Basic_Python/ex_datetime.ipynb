```python
import sys
def age_cal_eligibility(date1):
    try:
        cur_date = datetime.now()
        diff_year = cur_date.year - date1.year
        diff_mon = cur_date.month - date1.month
        flag = True
        if diff_year >= 21:

            print('eligible')

        elif diff_year==20 and diff_mon>=0:
            if cur_date.day - date1.day >=0:
                print("eligible")
            else:
                flag =False
                raise Exception("Not Eligible")
        else:
            flag = False
            raise Exception("Not eligible for this course")
    except:
        print(sys.exc_info()[1])
    return flag
```

```python
from datetime import date

def calculate_month_difference(start_date, end_date):
    """
    Calculates the difference in calendar months between two date objects.
    """
    return (end_date.year - start_date.year) * 12 + (end_date.month - start_date.month)

# Example usage:
date1 = date(2021, 12, 24)
date2 = date(2022, 3, 26)
months_diff = calculate_month_difference(date1, date2)
print(f"Difference in months: {months_diff}")
```
