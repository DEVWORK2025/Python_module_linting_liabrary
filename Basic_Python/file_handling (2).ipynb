```python

try:
    filename = "data.txt"
    total = 0
    count = 0

    with open(filename, 'r') as file:
        for line in file:
            numbers = line.split()
            for number in numbers:
                try:
                    num = float(number)
                    total += num
                    count += 1
                except ValueError:
                    print("Error: Invalid value found in the file.")

    if count > 0:
        average = total / count
        print("Average of numbers in the file:", average)
    else:
        print("No numbers found in the file.")

except FileNotFoundError:
    print("Error: File not found.")
```
